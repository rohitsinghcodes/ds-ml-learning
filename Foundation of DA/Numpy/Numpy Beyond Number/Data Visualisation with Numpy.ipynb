# Data Visualisation with Numpy

### Line Plot

Basic Line plot

```python
import numpy as np  # Importing NumPy library for numerical operations
import matplotlib.pyplot as plt  # Importing Matplotlib for plotting

x = np.linspace(0, 10, 100)  # Creating an array of 100 points from 0 to 10
y = np.cos(x)  # Calculating the cosine of each point in the array x

plt.plot(x, y)  # Plotting x vs y as a line plot
plt.xlabel('x-axis')  # Label for the x-axis
plt.ylabel('y-axis')  # Label for the y-axis
plt.title('Basic Line Plot')  # Title of the plot
plt.show()  # Display the plot
```

Custom Line Graph

```python
import numpy as np  # Importing NumPy library for numerical operations
import matplotlib.pyplot as plt  # Importing Matplotlib for plotting

x = np.linspace(0, 10, 50)  # Creating an array of 50 points from 0 to 10
y = np.cos(x)  # Calculating the cosine of each point in the array x

# Plotting x vs y with specified styling:
# 'o' denotes circular markers, linestyle is dashed ('--'), color is red,
# markersize is 5, and alpha (transparency) is set to 0.5
plt.plot(x, y, 'o', linestyle='--', color='red', markersize=5, alpha=0.5)

plt.xlabel('x-axis')  # Label for the x-axis
plt.ylabel('y-axis')  # Label for the y-axis
plt.title('Basic Line Plot')  # Title of the plot
plt.show()  # Display the plot
```

Line Plot with Multiple Lines

```python
import numpy as np  # Importing NumPy library for numerical operations
import matplotlib.pyplot as plt  # Importing Matplotlib for plotting

x = np.linspace(0, 10, 50)  # Creating an array of 50 points from 0 to 10
y1 = np.sin(x)  # Calculating the sine of each point in the array x
y2 = np.cos(x)  # Calculating the cosine of each point in the array x

# Plotting y1 (sin(x)) with specified styling: red dashed line, circular markers, 
# markersize 5, alpha 0.5, and label 'sin(x)'
plt.plot(x, y1, 'o', linestyle='--', color='red', markersize=5, alpha=0.5, label='sin(x)')

# Plotting y2 (cos(x)) with specified styling: blue dashed line, circular markers, 
# markersize 5, alpha 0.5, and label 'cos(x)'
plt.plot(x, y2, 'o', linestyle='--', color='blue', markersize=5, alpha=0.5, label='cos(x)')

plt.xlabel('x-axis')  # Label for the x-axis
plt.ylabel('y-axis')  # Label for the y-axis
plt.title('Basic Line Plot')  # Title of the plot
plt.legend()  # Displaying the legend to label sin(x) and cos(x)
plt.show()  # Display the plot
```

### Scatter Plot



Basic Scatter Plot

```python
x = np.random.rand(100)  # Generating 100 random x-coordinates
y = np.random.rand(100)  # Generating 100 random y-coordinates

plt.scatter(x, y)  # Creating a scatter plot of x vs y
plt.xlabel('x-axis')  # Label for the x-axis
plt.ylabel('y-axis')  # Label for the y-axis
plt.title('Basic Scatter Plot')  # Title of the plot
plt.show()
```

Customized Scatter Plot

```python
x = np.random.rand(100)  # Generating 100 random x-coordinates
y = np.random.rand(100)  # Generating 100 random y-coordinates
z = np.random.rand(100)  # Generating 100 random values for color mapping

plt.scatter(x, y, marker = 'o', c = z, alpha=0.7)  # Creating a scatter plot of x vs y
plt.xlabel('x-axis')  # Label for the x-axis
plt.ylabel('y-axis')  # Label for the y-axis
plt.title('Basic Scatter Plot')  # Title of the plot
plt.colorbar(label='Color Scale')  # Adding a color bar to indicate the scale of z values
plt.show()
```

Scatter Plot Wtih Multiple Class

```python
x1 = np.random.rand(100)  # Generating 100 random x-coordinates
y1 = np.random.rand(100)  # Generating 100 random y-coordinates
x2 = np.random.rand(100)  # Generating 100 random x-coordinates
y2 = np.random.rand(100)  # Generating 100 random y-coordinates

plt.scatter(x1, y1, marker = 'o', alpha=0.7)  # Creating a scatter plot of x vs y
plt.scatter(x2, y2, marker = 'o', alpha=0.7)  # Creating a scatter plot of x vs y
plt.xlabel('x-axis')  # Label for the x-axis
plt.ylabel('y-axis')  # Label for the y-axis
plt.title('Basic Scatter Plot')  # Title of the plot
plt.colorbar(label='Color Scale')  # Adding a color bar to indicate the scale of z values
plt.show()
```

### Bar Graph

Basic Bar Graph

```python
x = np.array(['A', 'B', 'C', 'D'])  # Categories for the bar graph
y = np.array([10, 15, 7, 12])  # Corresponding values for each category

plt.bar(x, y, alpha=0.7)  # Creating a bar graph with specified color and transparency
plt.xlabel('Categories')  # Label for the x-axis
plt.ylabel('Values')  # Label for the y-axis
plt.title('Basic Bar Graph')  # Title of the plot
plt.show()
```

```python
data = np.random.rand(10, 5)  # Generating a 10x5 array of random values

plt.bar(np.arange(10), data[:, 0], alpha=0.7, label='Category 1')  # Bar for Category 1
plt.bar(np.arange(10), data[:, 1], bottom=data[:, 0], alpha=0.7, label='Category 2')  # Bar for Category 2 stacked on Category 1
plt.bar(np.arange(10), data[:, 2], bottom=data[:, 0] + data[:, 1], alpha=0.7, label='Category 3')  # Bar for Category 3 stacked on Category 1 and 2
plt.xlabel('X-axis')  # Label for the x-axis
```

Grouped Bar Graph

```python
data = np.array([[10, 20, 30], [15, 25, 35], [20, 30, 40], [25, 35, 45]])  # Sample data for grouped bar graph
x = np.arange(4)  # X positions for the groups
plt.bar(x, data[:, 0], width=0.2, alpha=0.7, label='Group 1')  # Bars for Group 1
plt.bar(x + 0.2, data[:, 1], width=0.2, alpha=0.7, label='Group 2')  # Bars for Group 2
plt.bar(x + 0.4, data[:, 2], width=0.2, alpha=0.7, label='Group 3')  # Bars for Group 3
plt.xlabel('X-axis')  # Label for the x-axis
plt.ylabel('Values')  # Label for the y-axis
plt.title('Grouped Bar Graph')  # Title of the plot
plt.show()
```

### Histogram

Basic Histogram

```python
data = np.random.randn(1000)  # Generating 1000 random values from a normal distribution
print("Mean to be around 0:", np.mean(data))
print("Std to be around 1:", np.std(data))
plt.hist(data, bins = 30, alpha=0.7)  # Creating a histogram with 30 bins
plt.show()  # Display the histogram
```

Cumulative Histogram

```python
data = np.random.randn(1000)  # Generating 1000 random values from a normal distribution
plt.hist(data, bins=30, alpha=0.7, cumulative=True)
plt.show()  # Display the cumulative histogram
```

Histogram With Multiple Data Set

```python
# Generating two datasets of 100,000 random numbers from a standard normal distribution
# data2 is shifted by 2 units compared to data1
data1 = np.random.randn(100000)
data2 = np.random.randn(100000) + 2

# Plotting the histogram for data1 with 100 bins and an alpha (transparency) value of 0.7
# 'label' is used to provide a legend label for data1
plt.hist(data1, bins=100, alpha=0.7, label='D1')

# Plotting the histogram for data2 with 100 bins and an alpha (transparency) value of 0.7
# 'label' is used to provide a legend label for data2
plt.hist(data2, bins=100, alpha=0.7, label='D2')

# Displaying the legend for the plot to distinguish between D1 and D2
plt.legend()

# Displaying the histogram plot
plt.show()
```

### BoxPlot

```python
data = np.random.randn(100, 5)

print(np.mean(data, axis=0))
print(np.median(data, axis=0))

plt.boxplot(data, lables=['D1', 'D2', 'D3', 'D4', 'D5'])  # Creating a box plot for the data with filled boxes
plt.show()  # Display the box plot
```

### Heatmap

```python
data = np.random.rand(10, 10)  # Generating a 10x10 array of random values

plt.imshow(data)  # Displaying the data as a heatmap 
plt.colorbar()  # Adding a color bar to indicate the scale
plt.show()  # Display the heatmap
```

### Pie Chart

```python
# Defining the labels for the pie chart (programming languages)
labels = ['Python', 'C++', 'Java', 'JavaScript', 'PHP']

# Defining the corresponding popularity sizes for each language
sizes = [21.3, 14.6, 9.2, 8.5, 7.7]

# Defining the explode values for each slice (to offset certain slices for emphasis)
explode = (0.13, 0.1, 0, 0, 0)

# Defining custom colors for each slice of the pie
colors = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue', 'lightpink']

# Plotting the pie chart with labels, colors, and percentages displayed
plt.title("Programming Language Popularity")  # Title for the pie chart
plt.pie(sizes, colors=colors, labels=labels, autopct='%1.0f%%', shadow=True, explode=explode, startangle=140)

plt.axis('equal')  # Ensuring the pie chart is drawn as a circle (equal aspect ratio)
plt.legend(title="Languages", loc="upper left")  # Adding a legend to label each programming language

plt.show()  # Displaying the pie chart
```
